```python
import numpy as np 
import math 
import matplotlib.pyplot as plt 
from scipy.constants import h


h_bar = h/(2*np.pi)            #Simplified plancks constant
T = 2                          # Decay lifetime rate
E = np.linspace(-10,10,10)    #Energy values of squared wave
gamma =  h_bar/T               #Full width half parameter 
E_r = 2                        # Energy value of 2 Joules

def squared_wave(E): 
    Wave = pow(h_bar,2)/((2*np.pi)*((pow(gamma,2)/4) + pow(E_r,2) + pow(E,2)))
    return Wave
#Squared wave Function 


# Calculate the squared wave values
squared_wave_values = [squared_wave(e) for e in E]


# Create the plot
plt.figure(figsize=(8, 4))  # Optional: Adjust the figure size
plt.plot(E, squared_wave_values , color='blue')  # Plot the function with a  color
plt.title('Plot of squared wave')  # Set the title
plt.xlabel('E')  # Label for the x-axis
plt.ylabel('Wave squared')  # Label for the y-axis
plt.show()


    


```
