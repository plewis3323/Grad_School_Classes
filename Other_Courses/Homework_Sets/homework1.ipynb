## Round-off errors adding numbers

As discussed in class, generic floating point numbers $x_r$ cannot be exactly represented in the binary format. The representation of these numbers in the computer memory $x_c$ is therfore not exact. One may model computer respresentations of $x_c$ as random numbers with the (narrow) distribution

$$
\rho(x_c) = 
\begin{cases}
    {1 \over 2 \varepsilon x_r}&\text{for } x_r - \varepsilon x_r < x_c < x_r + \varepsilon x_r \\
    0 & \text{otherwise}
\end{cases}
$$

where for double precision numbers $\varepsilon = 1\cdot 10^{-16}$. 

We perform an addition of two numbers $x_1$ and $x_2$, i.e. $z = x_1 + x_2$. The computer representations of the numbers are stochastic with distributions noted above, i.e. $\rho(x_1)$ and $\rho(x_2)$. We also assume that the distributions of $x_1$ and $x_2$ are independent of each other, so that $\rho(x_1,x_2) = \rho(x_1)\rho(x_2)$. Just like $x_1$ and $x_2$, the sum $z$ is also a random number with a distribution $\rho_2(z)$ which may be different than the distributions of $x_1$ and $x_2$

a) Show that the probability distribution of the sum $z$ is given by 

$$

$$

$$

$$

c) Show further that the standard deviation (i.e., the square root of the variance) of the sum $z$
