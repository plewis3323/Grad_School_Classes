## Fibonacci numbers 

The sequence of integer numbers, starting with $u_1 = 1$ and $u_2=1$, and defined by 
$$ 
\boxed{
u_{n+1} = u_{n-1}+u_{n}
}
$$ 

are called **Fibonacci Numbers**. These numbers have been known for more than 2000 years and play a role in a wide range of phenomena and application ranging from tree branching and the arrangement of florets in sun-flowers to algorithms in financial mathematics.

Write a `Python` function to test the relation 
$$
\sum_{i=1}^nu_i^2 = u_nu_{n+1}
$$ for $n=2,\ldots, 32$.

```python
import numpy as np
import math


def Sum_Method_1(N):
    ''' This function computes the LHS of the Fibonacci relation. 

    Parameters
    -----------
    N: type = integer; represents N elements that the function generates. Also used for computing the n = 2, ... ,32 (In computer: n' = 1, ..., 31) 
    positions of LHS Summation. 

    Properties
    ----------
    The sum starts at 0 to N-1 elemental positions of the LHS sum. Still equivalent analytically 1 to N positions.

    Returns
    ----------
    List
       Prints out list of n = 2, ... ,32 (In computer: n' = 1, ..., 31) cases of the LHS summation. 
    '''
    F = [1, 1]
    Sum_L = []
    Sum = 0
    for i in range(2, N):
        F.append(F[i-2] + F[i-1])
    for i in range(1, N):
        Sum += pow(F[i-1],2)
        Sum_L.append(Sum)
    return Sum_L



def Sum_Method_2(N):
    ''' This function computes the RHS of the Fibonacci relation. 

    Parameters
    -----------
    N: type = integer; represents N elements that the function generates. Also used for computing the n = 2, ... ,32 (In computer: n' = 1, ..., 31) 
    positions of the RHS product. 


    Properties
    ----------
    The product runs by multiplying n-1 to n positions, but is still equivalent to the analytical form, which involves n to n+1 positions. 


    Returns
    ----------
    List
       Prints out list of n = 2, ... ,32 (In computer: n' = 1, ..., 31) cases of the RHS product. 

    '''   

    
    F = [1, 1]
    for i in range(2, N):
        F.append(F[i-2] + F[i-1])

    Sum_L2 = []
    for i in range(1, N):
        SumL2 = F[i] * F[i-1]
        Sum_L2.append(SumL2)

    return Sum_L2


print("Method 1:", Sum_Method_1(32))
print("Method 2:", Sum_Method_2(32))
print("LHS = RHS:", Sum_Method_1(32) == Sum_Method_2(32)) 

    





```

The *golden mean*  

$$ 
\varphi = {1 \over 2}\left (1 + \sqrt{5} \right )
$$

is related to the ratios of subsequent Fibonacci numbers $S_n = u_{n+1}/u_n$. Show that

$$
\boxed{
S_n = 1 + {1 \over S_{n-1}}
}
$$

which is an *finite continued fraction*

$$
S_n = 1 + {1 \over 1 + {1 \over 1 + \ldots}}
$$

Show analytically _and_ numerically that

$$
S_\infty = \varphi \,.
$$

### Part A. 


- $S_n = u_{n+1}/u_{n}$ from identity: $u_{n+1} = u_{n} + u_{n-1}.$

- $S_n = \frac{u_{n} + u_{n-1}}{u_{n}} =  1 + \frac{u_{n-1}}{u_n}$.


- Using a dummy substitution: $n^{'} - 1 = n$;  $S_{n^{'} - 1} = \frac{u_{n^{'}}}{u_{n^{'}-1}}$


- ***Final 1st Proof:*** $S_{n}  =  1 + \frac{1}{S_{n - 1}}$










### Part B. 

Starting with: $S_n =  1 + \frac{u_{n-1}}{u_n}$;  
In the 2nd term, divide each term in the fraction by $u_{n-1}$. 

$S_n =  1 + \frac{1}{\frac{u_n}{u_{n-1}}}$;  
do another dummy index substitution of $n^{''} + 1 = n$ and using $u_{n+1} = u_{n} + u_{n-1}$. 


$S_n =  1 + \frac{1}{\frac{u_{n} + u_{n-1}}{u_{n}}} = 1 + \frac{1}{1 + \frac{u_{n-1}}{u_{n}}}$


From here steps can be repeated again and again to get **Final 2nd Proof:** $$
S_n = 1 + {1 \over 1 + {1 \over 1 + \ldots}}
$$


      










### Part C. (Analytical Golden Mean Proof) 


Start With: $S_{n}  =  1 + \frac{1}{S_{n - 1}}$.


if large Limit up to $\infty$ is taken then in all practicallity: 
$S_{\infty}  =  1 + \frac{1}{S_{\infty}}$.

By alegebraic manipulations by multiplying LHS and RHS by $S_{\infty}$ and rearrangement of terms: 
$S_{\infty}^{2} - S_{\infty} -1  = 0 $.


By using quadratic equation a = 1, b = -1, and c = -1 and taking positive root: 
$S_{\infty} = \frac{1 +  \sqrt{5}}{2} = {1 \over 2}\left (1 + \sqrt{5} \right )$. 

***Final 3rd Proof:**  $S_{\infty} = \varphi$








```python
#Part D. (Numerical golden Mean Proof) 
import math
import numpy as np 
import pandas as pd 




def fibb_GM(N): 

    ''' This function takes a certain limit N and computes S_n = u_(n+1)/u_n with the goal as higher limits, 
    are taken, then the golden mean can be approximated.

    Parameters
    -----------
    N: type = integer; represents the first N elements that the function generates, starting with n = 1,2 (In computer: n' = 0,1). Also used as 
    a summation limit to approximate the golden mean.


    Returns
    ----------
    Float
       Prints out the ratio of the last term and 2nd the second last term in the Fibonacci sequence for golden mean calculation. 

    '''   
    F = [1,1]
    
    for i in range(2, N): 
        F.append(F[i-2] + F[i-1]) 

    GM = F[N-1]/F[N-2] 

    return GM 


print("Golden mean of N = 5:", fibb_GM(5))
print("\nGolden mean of N = 25:", fibb_GM(25))
print("\nGolden mean of N = 64:", fibb_GM(64))
print("\nGolden mean of N = 150:", fibb_GM(150))
print("\nGolden mean of N = 250:", fibb_GM(250))
print("\nGolden mean of N = 500:", fibb_GM(500))
print("\nGolden mean of N = 1000:", fibb_GM(1000))



"""

As demonstrated, the golden mean is proved as higher order terms are taken into effect as the limit
The limit of the Fibonacci sequence. 

"""





```

## Hermite polynomials

The Hermite polynomials are solutions of the differential equation
$$
\frac{d^2}{dx^2} H(x) - 2x \frac{d}{dx}H(x) + (\lambda - 1)H(x) = 0,
$$
which arises for example by solving Schrödinger’s equation for a particle confined to move in
a harmonic oscillator potential. The first few polynomials are
$$
\begin{aligned}
H_0(x) &= 1,\\
H_1(x) &= 2x,\\
H_2(x) &= 4x^2 - 2,\\
H_3(x) &= 8x^3 - 12x,\\
H_4(x) &= 16x^4 − 48x^2 + 12.
\end{aligned}
$$

They fulfill the recursion relation
$$
H_{n+1}(x) = 2x H_n(x) − 2n H_{n−1}(x).
$$

Write a `Python` function that computes the above Hermite polynomials for different values of $n$ using the given recursion relation. Check your results against some selected closed-form expressions.

**Hint:** The Hermite polynomials are implemented, e.g., in `scipy.special` as `eval_hermite()`.

```python
import math
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import scipy.special as sp




def Hermite(N, x): 
    ''' This function generates the Hermite polynomials via list appending using the recursion relation: H_(n+1) = 2*x*H_n - 2*n*H_(n-1). 

    Parameters
    -----------
    N: type = integer; represents a specific hermitian polynomial of the list. 
    x: type = np.array; represents the grid space for the horizontal axis. 


    Returns
    ----------
    List Element
       Outputs the specific polynomial of choice inside the list. ex.) to get H_3 case then N = 4 because of N-1 access of list.  

    '''   
    H = [1, 2*x] 
    for i in range(1, N+1): 
        H.append(2*x*H[i] - 2*(i)*H[i-1])
    return H[N] 


#Test n = 3 case:  
x = np.linspace(0,100, 400)
x_1 = np.linspace(0,200, 400)
y_sp = sp.eval_hermite(3, x) 
y_cust = Hermite(3,x) 
plt.plot(x, y_sp, label = "$H_{3}$ scipy method")
plt.plot(x_1, y_cust, label = "$H_{3}$ custom method")
plt.title("Hermite polynomial of degree 3 Cases")
plt.xlabel("x")
plt.ylabel("$H_{3}(x)$")
plt.legend()
plt.grid()
plt.show()









```

## Bessel functions via recursion

There are two linearly independent solutions $Z(x)$ to Bessel's differential equation:
$$
x^2 \frac{d^2 Z}{dx^2} + x \frac{dZ}{dx} + \left(x^2 - \nu^2 \right)Z = 0 \,,
$$
referred to as the Bessel functions of the first ($Z=J_\nu$) and second kind ($Z=Y_\nu$). The two Bessel functions satisfy the same recurrence relation:
$$
\boxed{
Z_{\nu+1} + Z_{\nu-1} = \frac{2\nu}{x} Z_\nu\,.
}
$$

1. Write a `Python` program that computes the Bessel function $J_\nu(x)$ for integer values $0 \leqslant \nu \leqslant 25$ and $x = 0.1, 1.0,$ and $10$ using upward and downward recursion.
2. Compare the two recursion methods with results obtained using `scipy.special.jv` (i.e, $J_\nu^\mathrm{(ref)}$) in a table, with the columns (in arbitrary order) $\nu$, $J_\nu^\mathrm{(ref)}$, $J_\nu^\mathrm{(up)}$, $J_\nu^\mathrm{(down)}$, and the relative difference
$$
\Delta_\nu = \frac{|J_\nu^\mathrm{(up)} - J_\nu^\mathrm{(down)}|}{|J_\nu^\mathrm{(up)}| + |J_\nu^\mathrm{(down)}|} \,.
$$
4. The errors for upward recursion depend on $x$; for certain values of $x$, both upward and downward recursions give similar results. Explain the reason for this and what it tells about your program.

```python
import math
import numpy as np 
import pandas as pd 
import scipy.special as sp




def J_up(N, x): 
    J_up = [sp.jv(0,x), sp.jv(1,x)] 
    for i in range(1, N+1): 
        J_up.append((2*(i)/x)*J_up[i] - J_up[i-1])
    return J_up[N] 
    

def J_down(N,x): 
    J_D = [sp.jv(N+1,x), sp.jv(N,x)]
    for i in range(N, 0, -1): 
        J_D.append((2*(i)/x)*J_D[-1] - J_D[-2])
    J_D1 = J_D[::-1]
    return J_D1[N] 


def Bessel_Table_UP(N, xL): 
    Dict = {} 
    for x in xL: 
        freq = [] 
        for k in range(N+1): 
            freq.append(J_up(k, x))
        Dict[f"x={x}"] = freq
    return pd.DataFrame(Dict, index=range(N+1))

def Bessel_Table_D(N, xL): 
    Dict = {} 
    for x in xL: 
        freq = [] 
        for k in range(N+1): 
            freq.append(J_down(k, x))
        Dict[f"x={x}"] = freq
    return pd.DataFrame(Dict, index=range(N+1))

print("Upward Bessel's:\n")
print("v_n", Bessel_Table_UP(25, [0.1, 1.0,10.0]))
        
print("Downward Bessel's:\n")
print("v_n", Bessel_Table_D(25, [0.1, 1.0,10.0]))    
    
    






```

```python
import math
import numpy as np 
import pandas as pd 
import scipy.special as sp


def Bessel_Dict(N,x): 
    Table = [] 
    for i in range(N+1): 
        J_ref = sp.jv(i,x)
        J_UP = J_up(i, x)
        J_D =  J_down(i,x)
        rel_diff = abs(J_UP - J_D)/(abs(J_UP) + abs(J_D))
        Table.append({
            "v_n": i, 
            "J_v(ref)": J_ref, 
            "J_v(up)": J_UP, 
            "J_v(down)": J_D,
            "Delta": rel_diff
        })
    return pd.DataFrame(Table)

print("Bessel Table at x = 10:\n", Bessel_Dict(25, 10))
            
        
        
        
    
```
