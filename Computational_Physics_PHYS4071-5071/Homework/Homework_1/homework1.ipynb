$

$

The value in decimal can be expressed as:

$
(\text{W})_{10} = -a_{N-1} 2^{N-1} + \sum_{i=0}^{N-2} a_i 2^i

Useful Property: 
$
    a_i = A (\text{since ai's are always 1.})

$

For \(r = 2\):

$

$

-----------------------------------------------------

### Applying to \(N = 32\):

$
= -2^{31} + \frac{1 - 2^{31}}{1 - 2}

$

**Final Answer:**

Write a `Python` function that converts a 2-byte signed integer number from the binary to the decimal representation and vice versa. Test it with $1100001010101101_2$.

```python
import numpy as np

#Binary to Decimal Conversion Program 
def binary_to_Decimal(b, signed=True): 
    bit = np.array(list(b), dtype = np.short)
    Power = np.array([2**15, 2**14, 2**13, 2**12, 2**11, 2**10, 2**9, 2**8, 2**7, 2**6, 2**5, 2**4, 2**3, 2**2, 2**1, 2**0])
    Decimals = 0
    if signed: 
        Decimal_0 = bit[0]*Power[0]*-1
    for k in range(1,len(Power)): 
            Decimals += bit[k]*Power[k]
    Final_Decimal = Decimals + Decimal_0
    return Final_Decimal

b = "1100001010101101"
Result = binary_to_Decimal(b)
print(f"{b}_2: {Result}_10")


#Decimal to binary Code (Unsigned) 
def Decimal_To_Binary_Unsigned(b): 
    Value = b
    binary = ""
    width = 16
    while Value > 0: 
        remainder =  Value%2
        binary = str(remainder) + binary
        Value = Value //2 
    if len(binary) == width: 
        print("Good Job") 
    elif len(binary) < width: 
        zeros_to_place = abs(width - len(binary))
        binary = "0"*zeros_to_place + binary  
    else: 
        print("No operations Take Place")
    return binary 

#Decimal to Binary (Signed)
def Decimal_To_Binary_Signed(b): 
    Value_A = -1*b 
    Value_B = Decimal_To_Binary_Unsigned(Value_A) 
    Un_S = np.array(list(Value_B))
    for p in range(len(Un_S)): 
        if Un_S[p] == "1": 
            Un_S[p] = "0"
        else: 
            Un_S[p] = "1" 
            

    add_1 = 1
    for L in range(len(Un_S) -1, -1, -1): 
        if Un_S[L] == "0" and add_1 == 1: 
             Un_S[L] = "1"
             add_1 = 0
        elif Un_S[L] == "1" and add_1 == 1: 
            Un_S[L] = "0"
            add_1 =  1 
    Var =  "".join(Un_S)          
    return Var
    

b_2 = -15699
Result_2 = Decimal_To_Binary_Signed(-15699)
print(f"{b_2}_10: {Result_2}_10")




            
    
    


```

$$ x_{1,2} = {- b \pm \sqrt{b^2 -4ac} \over 2a} $$

```python
# Quadratic Formula (Part 3a.)

import math
import numpy as np 

def Quad(a,b,c): 
    D = math.sqrt(pow(b,2) - 4*a*c)
    x_1 = (-b + D)/(2*a)
    x_2 = (-b - D)/(2*a)
    return x_1, x_2

Roots = Quad(1, -1*(10E6 + 10E-6), 1)
print("Roots (X) are:", Roots)
```

## Round-off errors adding numbers

As discussed in class, generic floating point numbers $x_r$ cannot be exactly represented in the binary format. The representation of these numbers in the computer memory $x_c$ is therfore not exact. One may model computer respresentations of $x_c$ as random numbers with the (narrow) distribution

$$
\rho(x_c) = 
\begin{cases}
    {1 \over 2 \varepsilon x_r}&\text{for } x_r - \varepsilon x_r < x_c < x_r + \varepsilon x_r \\
    0 & \text{otherwise}
\end{cases}
$$

where for double precision numbers $\varepsilon = 1\cdot 10^{-16}$. 

We perform an addition of two numbers $x_1$ and $x_2$, i.e. $z = x_1 + x_2$. The computer representations of the numbers are stochastic with distributions noted above, i.e. $\rho(x_1)$ and $\rho(x_2)$. We also assume that the distributions of $x_1$ and $x_2$ are independent of each other, so that $\rho(x_1,x_2) = \rho(x_1)\rho(x_2)$. Just like $x_1$ and $x_2$, the sum $z$ is also a random number with a distribution $\rho_2(z)$ which may be different than the distributions of $x_1$ and $x_2$

a) Show that the probability distribution of the sum $z$ is given by 

$$
\rho_2(z) = \int_{-\infty}^\infty \rho(x_1)\rho(z - x_1)dx_1
$$

We define  
$Z = X_1 + X_2$  

probability of z is defined: 
$P_2(z) = P(z|x_1x_2)P(x_1,x_2)$ $(P(x_1,x_2) = P(x_1)P(x_2))$ (defintions also works for probability densities) 

also: 
$P_2(z) = \int_{-\infty}^{\infty}\rho_{2}(z)dz$ = $\int_{-\infty}^{\infty}\int_{-\infty}^{\infty}\rho(z|x_1x_2)\rho(x_1)\rho(x_2)dx_1d_x2$ 

Next, use this substitution: 
($\rho(z|x_1x_2) = \delta(z - (x_1 + x_2))$) 


$P_2(z) = \int_{-\infty}^{\infty}\int_{-\infty}^{\infty}\delta(z - (x_1 + x_2))\rho(x_1)\rho(x_2)dx_1dx_2$

Apply the delta function to $x_2$-integration: 
$\int_{-\infty}^{\infty}\delta(z - (x_1 + x_2))\rho(x_2)dx_2 = \rho(z - x_1)$


**Final answer:**
$P_2(z) = \int_{-\infty}^{\infty}\rho(x_1)\rho(z - x_1)dx_1$










b) Show that the mean value of the sum is given by the sum of the mean values, i.e. 

$$
<z> = <x_1> + <x_2>
$$

We define $Z = X_1 + X_2$. The mean is  

$\langle Z \rangle = \int z \, \rho_2(z)\,dz.$  

Using marginalization, conditional relations ($\rho(x_1,x_2)$), and delta function substitution of $(\rho(x_1,x_2) = \delta(z - (x_1 + x_2)))$:  

$\rho_2(z) = \iint \delta(z - (x_1 + x_2)) \, \rho(x_1,x_2)\, dx_1 dx_2.$  

Therefore,  

$\langle Z \rangle = \int z \left( \iint \delta(z - (x_1+x_2)) \, \rho(x_1,x_2)\, dx_1 dx_2 \right) dz.$  

interchange the integrals:  

$\langle Z \rangle = \iint \left( \int z \, \delta(z - (x_1+x_2)) \, dz \right) \rho(x_1,x_2)\, dx_1 dx_2.$  

Using the sifting property of the delta function,  

$\int z \, \delta(z - (x_1+x_2)) \, dz = x_1 + x_2.$  

Thus,  

$\langle Z \rangle = \iint (x_1 + x_2)\, \rho(x_1,x_2)\, dx_1 dx_2.$  

Splitting the integral,  

$\langle Z \rangle = \iint x_1 \rho(x_1,x_2)\, dx_1 dx_2 + \iint x_2 \rho(x_1,x_2)\, dx_1 dx_2.$  

Therefore,  

$\langle Z \rangle = \langle x_1 \rangle + \langle x_2 \rangle.$
